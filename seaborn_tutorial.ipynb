# Seaborn

Seaborn is a Python library built on top of Matplotlib that focuses on statistical data visualization. It provides high-level functions, built-in themes, and automatic handling of datasets, allowing users to create informative and visually appealing plots with minimal code. Seaborn is widely used for exploring trends, relationships and distributions in data.

How Seaborn is better than Matplotlib?
1. Syntax ---> Seaborn is simpler ---> less lines of code is required whereas in Matplotlib a lot of Lines of code are required for bringing customisation.
2. By default ---> Seaborn Plots looks classic and that also without even bringing any additional customisation but in Matplotlib by default its plots looks similar to those of pandas plots.
3. plot types---> Matlolib ---> General plots(Line plot, Scatter, bar, histogram, pie) but in Seaborn ---> we'll have all of the visuals for statistical analysis (Kde, box, Regression Plot)
4. Integration ---> Matplotlib integrates with all data types while Seaborn integrates with Pandas Dataframe
5. Seaborn ---> More information we can extract from the plots or visuals but as compared to Matplotlib
6. A lot of new plots are supported in Seaborn that are not supported in Matplotlib

### Two Types of Graph
1. Figure Level Graph: When we create a Figure Graph then the container in which the graph is created that CONTAINER ---> Figure & that graph inside the container ---> Axis

The figure and axis can both be referred as an object only just like xLabel, yLabel, etc. are also objects.


There are 6 class of plots in Seaborn:

1. Relational ---> a. Scatter b. Lineplot

2. Distributional ---> a. Histogram b. KDE c. RugPlot

3. Categorical ---> a. Barplot b. violin c. Countplot d. Boxplot e. swarmplot

4. MAtrix ---> a. Heatmap b. clustermap

5. Regression Plot ---> regplot

6. Mutliplot ---> a. joint plot b. pairplot

### 1. HeatMap
#### (In every classification Machine Learning problem statement, we use the HeatMap.)

#### HeatMap ---> uses color shadesto show the size of numbers in a matrix 2D table ---> Darker or lighter color indicates higher or lower values.
#### superuseful when we want to see correlation, missing data, comparison, density

Example: Sales over the Month in different countries

For coming up with an Heatmap which we can convert into an Heatmap, we create a pivot table and then that pivot table only we can use to create an heatmap

#### pivot() Parameters:
##### index
##### columns
##### values


In matrix plot we do not have figure level functions, we only have axis level functions

#### heatmap() --->
##### Parameters:
##### data ---> matrix that has to be converted into an heatmap
##### annot ---> shows the numbers as well for the given rectangular grid
##### cmap ---> A specific styling/theme for the heatmap itself ---> coolwarm, virdis, etc.

```python
gap.head()
```

```python
dt=gap.pivot(index='country', columns='year', values='lifeExp')
```

```python
sns.heatmap(data=dt)
plt.show()
```

```python
dt=gap[gap['continent']=='Europe'].pivot(index='country', columns='year',values='lifeExp')
```

```python
sns.heatmap(data=dt)
plt.show()
```

```python
sns.heatmap(data=dt, annot=True)
plt.show()
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(data=dt, annot=True)
plt.show()
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(data=dt, annot=True, cmap='coolwarm', linewidth='0.5')
plt.show()
```
