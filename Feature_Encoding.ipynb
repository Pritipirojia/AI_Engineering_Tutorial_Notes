## Feature Scaling

---> we want to keep all the column values in a similar range scale.

### Feature scaling ---> It is the final step of feature engineering where we scale all of the column values on a similar scale range

#### Types of feature scaling:
*   Standardization
*   Normalization



### Standardization ---> It is the type of feature scaling where the scaling is done to keep all of the values in a standard scale range

Standard scale range ---> mean=0 & standard deviation=1

standardization of value= (value-mean)/standard deviation

scikitlearn ---> standardscaler ---> fit_transform

### Normalization ---> values must be scaled down to normal range ---> All the values must be in range of [0,1] ---> Min=0,Max=1 ---> use MinMaxScaler


Normalization of value: (value-min)/(max-min)

### Standardization vs Normalization

*   Standardization is optimal when you are going with distance based algorithm
*   If your data consists of outliers, then standardization is better than normalization. If you your data has too many outliers and if you don't want to remove it, you can standardize the data.
*   Normalization is mostly used for neural networks (deep lerning)
*   In normalization, huge issue will occur if data consists of outliers as in this case minmaxscaler won't provide good overall result
*   Apply normalization where scale uniformity matters like data preprocessing, deep learning for models like tensorflow, pytorch

## Column Transformer

This transformer can help us transforming ordinal data, nominal data & numeric data for the independent features without separating the columns from each other.

Parameters:


1.   Name that we provide the current transformer we are using
2.   Original or exact transformer ---> OrdinalEncoder or OneHotCoder or SimpleImputer
3.   Column_name ---> on which column you want to apply transformation


remainder=passthrough ---> it is not applicable for all the numerical column but only for columns which are left out.

```python
df=pd.read_csv('Bank_Marketing_Original_Data.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
df.columns
```

```python
transformer=ColumnTransformer(
    transformers=[
        ('tr1',OrdinalEncoder(categories=[['unknown','primary','secondary','tertiary']]),['Education']),
         ('tr2',OneHotEncoder(),['Job','Marital','Default','Housing','Loan','contact','Month','poutcome'])]
        ,remainder='passthrough')
```

```python
transformer
```

```python
x=df.iloc[:,:-1]
```

```python
x
```

```python
y=df['TARGET']
```

```python
y
```

```python
x=transformer.fit_transform(x)
```

```python
x
```

```python
le=LabelEncoder()
y=le.fit_transform(y)
```
