Boolean Index ---> False ---> Condition is NOT True

```python
avg
```

```python
### which youtubers are having average video length greater than 24 minutes
avg > 24
```

## Plotting graphs in Pandas

For plotting graphs in Pandas, we require plot() ---> df.plot()

```python
avg
```

```python
avg.plot()
```

kind ---> this parameter decides what kind of graph you want to plot

```python
dataframe.sort_values(ascending=False).head(15)
```

```python
dataframe.sort_values(ascending=False).head(15).plot(kind='bar')
```

```python
dataframe.sort_values(ascending=False).sample(15).plot(kind='bar')
```

```python
dataframe.sort_values(ascending=False).sample(15).plot(kind='pie')
```
