## Exploratory Data Analysis

For deep learning, feature engineering is not required as its algo is itself capable of detecting features on its own. But for ML model, EDA s required as ML algo can't extract features from data.

DL algo are great with huge amount of data but ML algo can be built with small amount of data.

For ML model building, we need to know which feature is important and which feature is not important.

### Steps for EDA

##### Categorise the given column as a numerical, categorical, mixed data type column
#### Univariate Analysis ---> Numerical column or Categorical column
##### Numerical column Analysis ---> Distribution of the Data ---> Histogram, kdeplot
##### ---> outliers ---> Boxplot
##### ---> Handle Missing values

##### Categorical Column Analysis ---> Distribution of the Data ---> countplot ---> Missing values --->

#### Bivariate Analysis ---> Numerical Numerical ---> scatterplot
##### Numerical categorical ---> barplot
##### categorical Categorical ---> Heatmap

##### EDA ---> Feature Engineering

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('tested.csv')
```

```python
df.head()
```

#### Univariate Analysis on Numerical Column --->
#### Data Distribution of the Numerical Column ---> describe() & for plotting ---> kdeplot or histogram
#### outlier Detection ---> Boxplot

```python
df['Age'].describe()
```

```python
df[df['Age']==76]
```

```python
sns.boxplot(data=df, x='Age')
```

```python
df['Age'].isnull().sum()
```

```python
## As there are outlers 76 and 65 in age column, so can fill the column using median and not mean
df['Age']=df['Age'].fillna(df['Age'].median())
```

```python
df['Age'].isnull().sum()
```

```python
df['Age'].describe()
```

```python
sns.histplot(data=df,x='Age')
```

```python
sns.kdeplot(data=df,x='Age')
```

```python
df.head()
```

```python
df['Fare'].describe()
```

```python
df[df['Fare']==512.329200]
```

```python
sns.histplot(data=df,x='Fare')
```

```python
sns.histplot(data=df,x='Fare',bins=20)
```

#### Categorical Categorical ---> crosstab() ---> two categories as input on which we can perform the analysis

```python
pd.crosstab(df['Pclass'],df['Survived'])
```

```python
pd.crosstab(df['Pclass'],df['Survived'],normalize=True)
```

```python
pd.crosstab(df['Pclass'],df['Survived'],normalize=True)*100
```

```python
sns.heatmap(pd.crosstab(df['Pclass'],df['Survived'],normalize=True)*100)
```
