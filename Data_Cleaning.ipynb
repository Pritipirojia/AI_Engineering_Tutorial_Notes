### Automated Data Cleaning

#### Types of missing values
1.   Missing completely at Random (MCAR) ---> No pattern is there for missing data as could be due to system glitch or the data entry guy might would have skipped the value
2.   Missing at Random (MAR) ---> Pattern is there for missing data
3.   Missing Not at Random (MNAR) ---> if the missing values are there because of deliberate reasons but no pattern is there

#### Ways of cleaning missing data

1. Drop missing data ---> dropna()

2. Mean / Meadian Imputation ---> Fill the missing data with the MEan or Median of all rows of that column

while going with mean value ---> check of the outliers ---> if there are outliers, use median



```python
df['Heart Rate (BPM)'].describe()
```

```python
df['Heart Rate (BPM)'].mean()
```

```python
df['Heart Rate (BPM)'].median()
```

```python
### As there is bigger difference between 75% value and max so we will use median to fill the null values
df['Heart Rate (BPM)']=df['Heart Rate (BPM)'].fillna(df['Heart Rate (BPM)'].median())
```

```python
df['Heart Rate (BPM)'].isnull().sum()
```

Different Imputation Techniques:-
1. Remove the missing values
2. Mean/MEdian Imputation ---> For numeric columns
3. Mode Imputation ---> For categorical column
4. constant or sentinal value Imputation
5. Forward filling or Backward Filling
6. Simple Imputation ---> simple imputer from scikit learn ---> it takes input as 2D-vector

```python
df=pd.read_csv('unclean_smartwatch_health_data.csv')
```

```python
df.info()
```

```python
df=df.drop(columns=['User ID'])
```

```python
from sklearn.impute import SimpleImputer
imputer=SimpleImputer(strategy='median')
```

```python
df['Heart Rate (BPM)']=imputer.fit_transform(df['Heart Rate (BPM)'])
```

```python
df['Heart Rate (BPM)']=imputer.fit_transform(df[['Heart Rate (BPM)']])
```

```python
df['Heart Rate (BPM)']=imputer.fit_transform(df[['Heart Rate (BPM)']]).ravel()
```

```python
df.info()
```

```python
imputer=SimpleImputer(strategy='mean')
```

```python
df['Blood Oxygen Level (%)']=imputer.fit_transform(df[['Blood Oxygen Level (%)']]).ravel()
```

```python
df.info()
```

```python
imputer=SimpleImputer(strategy='most_frequent')
df['Activity Level']=imputer.fit_transform(df[['Activity Level']]).ravel()
```

```python
df.info()
```

```python
imputer=SimpleImputer(strategy='constant',fill_value=5)
df['Stress Level']=imputer.fit_transform(df[['Stress Level']]).ravel()
```

```python
df.info()
```
