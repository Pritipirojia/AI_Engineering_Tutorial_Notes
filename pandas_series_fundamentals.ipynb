## Pandas

Pandas is a powerful library used for data manipulation and analysis. It helps us to work with data in Tabular format, just like how we deal with data and spreadsheet in excel or data.

Pandas provides functionalities and attributes to clean, organize and analyze large amounts of data

1. Easy to use---> Works like Excel inside python
2. Fast---> Handles large data faster than manual methods
3. Clean data---> helps clean, filter and transform messy data
4. Integrates well---> Matplotlib, Seaborn
5. Ideal for data science, ML and AI

```python
import numpy as np
import pandas as pd
```

Pandas series is a column of data in a table like in excel or google sheets with labels called as indexes attached to each value

```python
pd.Series(data=[40,50,60],index=['Bread', 'Milk', 'Butter'], name='Price of grocery items')
```

Whenever you create a series, data type is always going to be decided on the basis of the values itself that are there in the series

Series is a 1D array that can hold any type of data like integers, float, string (objects), boolean, mixed data type.

## Creating a series in multiple ways
1. List
2. Numpy array
3. Dictionary
4. custom index
5. scalar value

```python
## series from list
country=['India', 'china', 'USA', 'Pakistan']
pd.Series(country)
```

```python
batsman_runs=[48,64,73,98,65,90]
pd.Series(batsman_runs, dtype='int32')
```

```python
subject_names=['Maths', 'Physics', 'Chemistry', 'English']
subject_marks=[48,92,67,98]

pd.Series(data=subject_marks, index=subject_names, name='Subject Marks data', dtype='int32')
```

```python
### Creating an series using a numpy array
values=np.arange(10,21)
pd.Series(values)
```
