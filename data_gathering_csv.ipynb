### read_csv() Parameters

1.   sep (separation) ---> how values are separated from each other in a given file
csv---> amit,28 ---> sep=','
tsv---> amit\t28 ---> sep='\t'




```python
pd.read_csv('demo_mlexpec.tsv',sep='\t')
```


2.   names ---> names is going to take a list of names that we can provide to our column names

---> names=['Name','Age']





```python
pd.read_csv('titanic.csv')
```

```python
pd.read_csv('titanic.csv',names=['P_ID','Survived','Pclass','P_Name','Sex','Age','Sibling or Spouse','Parch','Ticket_ID','Fare','Cabin','Starting Journey'])
```

3.   header ---> header parameter makes sure that any of the row name will be shown as the column name only based on indexing of rows




```python
pd.read_csv('titanic.csv')
```

```python
pd.read_csv('titanic.csv',header=1)
```

4.   index_col ---> keep one of our column as the index



```python
pd.read_csv('titanic.csv',index_col='PassengerId')
```

5.   usecols ---> what columns we should usw to show out of all of the columns we are having in the dataframe ---> uses a list to mention all of the column names which are required



```python
pd.read_csv('student_depression_dataset.csv')
```

```python
pd.read_csv('student_depression_dataset.csv', usecols=['id','Gender',	'Age','City','Profession','Academic Pressure','CGPA','Study Satisfaction','Sleep Duration','Dietary Habits','Degree','Have you ever had suicidal thoughts ?','Work/Study Hours','Depression'])
```

6.   squeeze ---> DataFrame ---> Series ---> squeeze ---> combination of usecols and squeeze




```python
pd.read_csv('titanic.csv',usecols=['Survived']).squeeze()
```

7.   nrows ---> it is going to help us with a fixed number of rows out of the complete data ---> 100 ---> only 100 rows out of all rows will be shown




```python
pd.read_csv('student_depression_dataset.csv',nrows=150)
```

### Encoding issue ---> utf-8 ---> e.g.: in case of some text data or emoji data ---> manually keep the encoding to utf-8 so that we can read the content from the dataset



8.   encoding ---> It makes sure the data we are reading is in utf-8 format ---> encoding ='latin-1'

10.   dtype ---> It can assign initially only with a specific data type to a column ---> dtype={'col_1':int,'col_2':float}

12.  converters ---> bring the transformation on a specific value in a column to another value & this converter is a function

converters={'col_name':function}

```python
def station(value):
  if value=='Q':
    return 'Queenstown'
  else:
    return value
```

```python
pd.read_csv('titanic.csv',converters={'Embarked':station})
```

13. na_values ---> can help us with mentioning what values to be considered as a missing value

na_values=['',':','No']

```python
pd.read_csv('titanic.csv',na_values=['\','No',NaN])
```
