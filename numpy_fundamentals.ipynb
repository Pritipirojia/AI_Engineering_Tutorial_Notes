## Stacking and Splitting in Array --> size of the arrays must be the same

## Stacking ---> Joining the arrays together
## Horizontal Stacking ---> Row by Row joining is done --> hstack()
## Vertical Stacking ---> Column by Column Joining is done ---> vstack()


```python
array1=np.arange(1,13).reshape(3,4)
```

```python
array1
```

```python
array2=np.arange(13,25).reshape(3,4)
```

```python
array2
```

```python
horizontal_array=np.hstack((array1,array2))
```

```python
horizontal_array
```

```python
horizontal_array=np.hstack((array1,array2,array1))
```

```python
horizontal_array
```

```python
vertical_array=np.vstack((array1,array2))
```

```python
vertical_array
```
