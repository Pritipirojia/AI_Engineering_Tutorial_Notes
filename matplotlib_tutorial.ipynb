### Customization for line

Parameters:
1. color ---> what is going to be the color of the line
2. marker ---> how point of intersection appears
3. linestyle ---> it will be defining the style of the line ---> '--', '-.'
4. linewidth ---> shows the width of the line
5. markersize ---> shows the markersize or it can change the size of the marker, default=6
6. label ---> label that what this specific line is representing but is only helpful while plotting multiple lines in a single visualisation

### Stacked Bar chart

Instead of keeping the bars in side by side direction, we can keep the bars in stacked manners i.e. One on top of other in order to figure out which category has maximum sale by looking over the length of the bar

bar()---> parameter---> bottom---> This bottom parameter specifies that what bar i want to keep below my current bar

As we want to differentiate on the basis of bar length value so that is why we have to keep y value of previous bar at the bottom


```python
plt.figure(figsize=(7,4))
plt.title('Month Based Total Boxes shipped')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.bar(m.index,m.values, color='purple', label='MintChip Choco')
plt.bar(d.index,d.values, color='pink', label='85% Dark Bars', bottom=m.values)
plt.bar(p.index,p.values, color='violet', label='Peanut Butter Cubes', bottom=m.values + d.values)
plt.xticks(np.arange(m.shape[0]),x)
plt.legend()
plt.show()
```
