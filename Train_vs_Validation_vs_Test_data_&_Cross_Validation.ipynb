## Train vs Validation vs Test data & Cross Validation

### Train data ---> it is that pattern of the data that helps model to learn pattern in between input & output

*   Training data model sees the data during training.
*   Training data is used to fit the model
*   It is the representation of problem domain


Training ---> model learns how to map output and input by deriving an mathematical equation

##### Hyperparameter tuning ---> the process of finding the optimal set of hyperparameters for a machine learning model to maximize performance while avoiding overfitting or underfitting.

##### Hyperparameters ---> Unlike model parameters learned during training, hyperparameters are set before training and control aspects like learning rate, regularization strength, or tree depth.

### Validation data ---> Data on which model does not learn the pattern. Used for optimal model selection, hyperparameter tuning and early stopping. It helps to detect overfitting by checking performance on unseen data during training.

### Test data ---> completely unseen data by the model during training or tuning, no output is provided. A final hold-out dataset. It is used to evaluate model performance objectively. Provides final unbiased evaluation before deploying in real world.

Training data ---> Model learns pattern
Validation data ---> Model learns hyperparameters
Test data ---> Model predicts output

### Cross-validation ---> technique used to check how well a machine learning model performs on unseen data while preventing overfitting. It works by:

*   Splitting the dataset into several parts.
*   Training the model on some parts and testing it on the remaining part.
*   Repeating this resampling process multiple times by choosing different parts of the dataset.
*   Averaging the results from each validation step to get the final performance.
